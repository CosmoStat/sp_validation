```python
R_shear_ind = {}
for sh in shapes:
    # Shear response per galaxy
    R_shear_ind[sh] = gal_metacal[sh].R_shear
```

### Write basic shape catalogue

```python
local_cal = False

if local_cal:
    # MKDEBUG TODO: add local selection matrix
    for sh in shapes:
        write_shape_catalog(
            f'{output_shape_cat_base}_{sh}.fits',
            ra[sh],
            dec[sh],
            g_corr_mc[sh],
            w[sh],
            mag[sh],
            gal_metacal[sh].R,
            R_shear[sh],
            gal_metacal[sh].R_selection,
            c[sh],
            c_err[sh],
            alpha_leakage=alpha_leak_mean[sh],
            g1_uncal=g_uncorr[sh][0],
            g2_uncal=g_uncorr[sh][1], 
            R_g11=R_shear_ind[sh][0, 0],
            R_g12=R_shear_ind[sh][0, 1],
            R_g21=R_shear_ind[sh][1, 0],
            R_g22=R_shear_ind[sh][1, 1],
            add_cols=add_cols_data[sh],
        )
else:
    for sh in shapes:
        
        write_shape_catalog(
            f'{output_shape_cat_base}_{sh}.fits',
            ra[sh],
            dec[sh],
            g_corr_mc[sh],
            w[sh],
            mag[sh],
            gal_metacal[sh].R,
            R_shear[sh],
            gal_metacal[sh].R_selection,
            c[sh],
            c_err[sh],
            alpha_leakage=alpha_leak_mean[sh],
            g1_uncal=g_uncorr[sh][0],
            g2_uncal=g_uncorr[sh][1],
            add_cols=add_cols_data[sh],
        )
```

### Write extended shape catalogue

```python
for sh in shapes:
    
    # Signa-to-noise estimates
    if sh == 'ngmix':
        my_snr = snr['ngmix']
    elif sh == 'galsim':
        my_snr = dd['SNR_WIN'][m_gal['galsim']]

    # Additional columns:
    # {e1, e2, size}_PSF
    ext_cols = {
        'e1_PSF' : dd[key_PSF_ell[sh]][:,0][m_gal[sh]][mask[sh]],
        'e2_PSF' : dd[key_PSF_ell[sh]][:,1][m_gal[sh]][mask[sh]],
        'fwhm_PSF' : size_to_fwhm[sh](dd[key_PSF_size[sh]][m_gal[sh]][mask[sh]])
    }

    # Extended catalogue with SNR, individual R matrices, ext_cols
    write_shape_catalog(
        f'{output_shape_cat_base}_extended_{sh}.fits',
        ra[sh],
        dec[sh],
        g_corr_mc[sh],
        w[sh],
        mag[sh],
        gal_metacal[sh].R,
        R_shear[sh],
        gal_metacal[sh].R_selection,
        c[sh],
        c_err[sh],
        snr=my_snr,
        alpha_leakage=alpha_leak_mean[sh],
        g1_uncal=g_uncorr[sh][0],
        g2_uncal=g_uncorr[sh][1],
        R_g11=R_shear_ind[sh][0, 0],
        R_g12=R_shear_ind[sh][0, 1],
        R_g21=R_shear_ind[sh][1, 0],
        R_g22=R_shear_ind[sh][1, 1],       
        add_cols=ext_cols,
     )
```
