```python
local_cal = False

if local_cal:
    for sh in shapes:
        
        # Toal response per galaxy
        R = np.zeros_like(gal_metacal[sh].R_shear)
        for i in (0,1):
            for j in (0, 1):
                R[i, j] = (
                    gal_metacal[sh].R_shear[i, j]
                    + gal_metacal[sh].R_selection[i, j]
                )
    
        # Signa-to-noise estimates
        if sh == 'ngmix':
            my_snr = snr['ngmix']
        elif sh == 'galsim':
            my_snr = dd['SNR_WIN'][m_gal['galsim']]

        write_shape_catalog(
            f'{output_shape_cat_base}_{sh}.fits',
            ra[sh],
            dec[sh],
            g_corr_mc[sh],
            w[sh],
            mag[sh],
            my_snr,
            gal_metacal[sh].R,
            R_shear[sh],
            gal_metacal[sh].R_selection,
            c[sh],
            c_err[sh],
            alpha_leak_mean[sh],
            g1_uncal=g_uncorr[sh][0],
            g2_uncal=g_uncorr[sh][1], 
            R_11=R[0, 0],
            R_12=R[0, 1],
            R_21=R[1, 0],
            R_22=R[1, 1],
            add_cols=add_cols_data,
        )
else:
    for sh in shapes:
        
         # Signa-to-noise estimates
        if sh == 'ngmix':
            my_snr = snr['ngmix']
        elif sh == 'galsim':
            my_snr = dd['SNR_WIN'][m_gal['galsim']]

        write_shape_catalog(
            f'{output_shape_cat_base}_{sh}.fits',
            ra[sh],
            dec[sh],
            g_corr_mc[sh],
            w[sh],
            mag[sh],
            my_snr,
            gal_metacal[sh].R,
            R_shear[sh],
            gal_metacal[sh].R_selection,
            c[sh],
            c_err[sh],
            alpha_leak_mean[sh],
            g1_uncal=g_uncorr[sh][0],
            g2_uncal=g_uncorr[sh][1],
            add_cols=add_cols_data[sh],
        )
```
